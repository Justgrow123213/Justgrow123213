# Пример скрапинга данных о недвижимости

В этом ноутбуке показано, как использовать скраперы для сбора данных о недвижимости с сайтов DDProperty и M2Agent.

```python
# Импортируем необходимые модули
import sys
import os
sys.path.append(os.getcwd())

from real_estate_agents.database.property_db import PropertyDatabase
from real_estate_agents.data_collector.agent import DataCollectionAgent
from real_estate_agents.data_collector.scrapers import DDPropertyScraper, M2AgentScraper
```

## Инициализация базы данных и агента сбора данных

```python
# Создаем базу данных и агента
db = PropertyDatabase(db_file="scraped_properties.json")
agent = DataCollectionAgent(db)
```

## Скрапинг данных с DDProperty

```python
# Скрапинг кондоминиумов в Бангкоке
bangkok_condos = agent.scrape_ddproperty(
    location="bangkok",
    property_type="condo",
    max_pages=1
)

print(f"Добавлено {len(bangkok_condos)} кондоминиумов в Бангкоке")
```

## Скрапинг данных с M2Agent

```python
# Скрапинг квартир в Москве
moscow_apartments = agent.scrape_m2agent(
    city="moscow",
    property_type="квартиры",
    max_pages=1
)

print(f"Добавлено {len(moscow_apartments)} квартир в Москве")
```

## Просмотр собранных данных

```python
# Получаем все объекты из базы данных
all_properties = db.properties
print(f"Всего объектов в базе: {len(all_properties)}")

# Выводим первые 3 объекта
for prop in all_properties[:3]:
    print(f"\nОбъект: {prop.get('property_type')} в {prop.get('location')}")
    print(f"Цена: {prop.get('price')}")
    print(f"Площадь: {prop.get('area')} {prop.get('area_unit')}")
    print(f"Спальни: {prop.get('bedrooms')}")
    print(f"Ванные: {prop.get('bathrooms')}")
    if 'features' in prop and prop['features']:
        print(f"Особенности: {', '.join(prop['features'][:3])}...")
    print(f"Источник: {prop.get('source')}")
```

## Поиск объектов по критериям

```python
# Поиск объектов по критериям
query = {
    'min_price': 100000,
    'max_price': 500000,
    'bedrooms': 2
}

matching_properties = db.search_properties(query)
print(f"Найдено {len(matching_properties)} объектов, соответствующих критериям")

# Выводим первые 3 объекта
for prop in matching_properties[:3]:
    print(f"\nОбъект: {prop.get('property_type')} в {prop.get('location')}")
    print(f"Цена: {prop.get('price')}")
    print(f"Площадь: {prop.get('area')} {prop.get('area_unit')}")
    print(f"Спальни: {prop.get('bedrooms')}")
    print(f"Источник: {prop.get('source')}")
```
